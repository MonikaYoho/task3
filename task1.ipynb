##list comprehension vs for loop

```python
numbers=list(range(10))
squares=[x*x for x in numbers]
print(squares)
```

```python
%timeit[x*x for x in numbers]
```

```python
def square_loop(number):
    result=[]
    for x in numbers:
        result.append(x*x)
        return result

        %timeit square_loop(numbers)
```
